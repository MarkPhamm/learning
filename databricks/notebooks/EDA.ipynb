```python
# Load the consumer complaints data
df = spark.read.table("workspace.consumer_complaints.jpmorgan_complaints_delta")

# Display the first few rows for a quick overview
display(df)
```

```python
# Visualize the number of complaints by product
complaints_by_product = df.groupBy("product").count().orderBy("count", ascending=False)
display(complaints_by_product)

# Visualize the number of complaints by state
complaints_by_state = df.groupBy("state").count().orderBy("count", ascending=False)
display(complaints_by_state)

# Visualize the number of complaints over time
complaints_by_date = df.groupBy("date_received").count().orderBy("date_received")
display(complaints_by_date)
```
